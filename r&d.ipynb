```python
def validate_test(text):
    if not text:
        raise ValueError("Input text cannot be empty")
    if not text.strip():
        raise ValueError("Input text cannot contain only blank spaces")

validate_test("     ")
```

```python
def validate_file_extension(filename):
    file_extension=Path(filename).suffix.lower()
    if file_extension not in ['.txt', '.pdf', '.docx']:
        raise ValueError(f"Unsupported file type: {file_extension}")

validate_file_extension("sample.PDF")
```

```python
def validate_document_metadata(metadata):
    if metadata is None:
        raise ValueError("Metadata cannot be None")
    if type(metadata) != dict:
        raise TypeError("Metadata must be a dictionary")
    for key in ['case_title', 'court', 'year']:
        if key not in metadata:
            raise KeyError(f"Metadata must contain '{key}'")
    if not metadata['case_title'] or not metadata['court']:
        raise ValueError("Metadata must contain 'case_title' and 'court'")
    if type(metadata['year']) != int:
        raise TypeError("Year must be an integer")
    return True


metadata = {
    "case_title": "ABC vs State of Maharashtra",
    "court": "Supreme Court of India",
    "year": 2024
}

validate_document_metadata(metadata)
```

```python
from pydantic import BaseModel,Field
from typing import Optional

class DocumentRequest(BaseModel):
    # Required AND non-empty (at least 1 character)
    document_type: str = Field(..., min_length=1)
    court: str = Field(..., min_length=1)
    
    # Required AND positive integer (strictly greater than 0)
    year: int = Field(..., gt=0)    
    # Optional (defaults to None), but if provided, must be non-empty
    case_title: Optional[str] = Field(default=None, min_length=1)

data = {
    "case_title": "ABC vs State of Maharashtra",
    "court": "Supreme Court of India",
    "year": 2024,
    "document_type": "judgment"
}

doc = DocumentRequest(**data)
print(doc)
```
